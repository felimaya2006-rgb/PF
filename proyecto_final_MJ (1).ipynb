Trece (13) características clínicas:

age: edad del paciente (años)

anaemia: anemia; disminución de glóbulos rojos o hemoglobina (booleano)

creatinine phosphokinase (CPK): nivel de la enzima CPK en la sangre (mcg/L)

diabetes: si el paciente tiene diabetes (booleano)

ejection fraction: fracción de eyección; porcentaje de sangre que sale del corazón en cada contracción (porcentaje)

high blood pressure: si el paciente tiene hipertensión (booleano)

platelets: plaquetas en la sangre (kiloplaquetas/mL)

sex: sexo del paciente, mujer u hombre (binario)

serum creatinine: nivel de creatinina sérica en la sangre (mg/dL)

serum sodium: nivel de sodio sérico en la sangre (mEq/L)

smoking: si el paciente fuma o no (booleano)

time: período de seguimiento (días)

death event: si el paciente murió durante el período de seguimiento (booleano)

Estas son las personas la cuales tienen más del máximo del número "normal" de plaquetas que son 450.000
estas personas se denominan con una enfermedadd extra la cúal es **TROMBOCITOCIS**,  y puede ser una respuesta a otra condición (trombocitosis reactiva) o ser causado por un problema primario en la médula ósea (trombocitosis primaria).

```python
data[data['platelets']>450000]
```

```python
def clasificar_plaquetas(value):
    if value < 150000:
        return "trombocitopenia"
    elif value > 450000:
        return "trombocitosis"
    else:
        return "rango normal"

dead['categoria_plaquetas'] = dead['platelets'].apply(clasificar_plaquetas)

conteo_plaquetas = dead['categoria_plaquetas'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(conteo_plaquetas.values,
        labels=conteo_plaquetas.index,
        autopct='%1.1f%%')
plt.title("Distribución de categorías de plaquetas en personas fallecidas")
plt.show()
```

se logro observar que la cantidad de plaquetas no afecta mucho en el fallecimiento de las personas

# Se decidió estudiar las variables externas que pueden influir en la enfermedad de las personas con estas fallas cardiacas las cuales son anemia, diabetes y si son o no fumadores

```python
anemia_counts = dead['anaemia'].value_counts().sort_index()
diabetes_counts = dead['diabetes'].value_counts().sort_index()
smoking_counts = dead['smoking'].value_counts().sort_index()

plt.figure(figsize=(10,6))

labels = ['No', 'Sí']

# Anemia
plt.subplot(1, 3, 1)
plt.bar(labels, anemia_counts)
plt.title("Anemia (Fallecidos)")
plt.ylabel("Número de Pacientes")

# Diabetes
plt.subplot(1, 3, 2)
plt.bar(labels, diabetes_counts)
plt.title("Diabetes (Fallecidos)")

# Fumadores
plt.subplot(1, 3, 3)
plt.bar(labels, smoking_counts)
plt.title("Fumadores (Fallecidos)")

plt.tight_layout()
plt.show()
```

```python
def clasificar_ef(value):
    if 55 <= value <= 70:
        return "Normal (55–70%)"
    elif 40 <= value < 55:
        return "Leve disminución (40–55%)"
    elif value < 40:
        return "Insuficiencia cardíaca (<40%)"
    else:
        return "Anormalmente alta (70–100%)"

dead['categoria_ef'] = dead['ejection_fraction'].apply(clasificar_ef)

conteo_ef = dead['categoria_ef'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(conteo_ef.values,
        labels=conteo_ef.index,
        autopct='%1.1f%%')
plt.title("Distribución de fracción de eyección en personas fallecidas")
plt.show()
```

El mapa de calor muestra qué variables están más relacionadas entre sí.
En este dataset, las variables que más se relacionan con la muerte son:
la edad, la creatinina sérica, la fracción de eyección y el tiempo de seguimiento.

1. DEATH_EVENT tiene correlación positiva con:

Edad (0.25) =  A mayor **edad**, mayor **probabilidad de muerte**.

Creatinina sérica (0.29) = La falla renal aumenta el riesgo de muerte.

2. DEATH_EVENT tiene correlación negativa con:

Fracción de eyección (-0.27)= Menor ejection fraction = peor funcionamiento del corazón = **mayor riesgo.**

Tiempo (-0.53) = “Time” significa cuánto sobrevivió el paciente.
Correlación negativa fuerte significa que:

A menor tiempo → mayor probabilidad de muerte.


```python
correlation_matrix = data.corr(numeric_only=True)

# Obtener la lista de columnas
cols_for_corr = correlation_matrix.columns.tolist()

# Crear la figura del mapa de calor
fig, ax = plt.subplots(figsize=(10, 8))
cax = ax.matshow(correlation_matrix, cmap='coolwarm')

# Barra de color
fig.colorbar(cax, label='Coeficiente de Correlación')

# Etiquetas
ax.set_xticks(np.arange(len(cols_for_corr)))
ax.set_yticks(np.arange(len(cols_for_corr)))
ax.set_xticklabels(cols_for_corr, rotation=90)
ax.set_yticklabels(cols_for_corr)

# Añadir los valores numéricos dentro de la matriz
for i in range(len(cols_for_corr)):
    for j in range(len(cols_for_corr)):
        ax.text(j, i, f'{correlation_matrix.iloc[i, j]:.2f}',
                ha="center", va="center", color="black", fontsize=8)

# Título
plt.title('Mapa de Calor de Correlación - Datos de Insuficiencia Cardíaca', pad=20)

# Mostrar
plt.show()
```

Segun el mapa de calor podemos observar que la creatinina alta tiene una fuerte relación al fallecimiento del paciente por lo tanto se creó un gráfico de bigotes donde lo vemos con máyor claridad que la creatinina más alta es de los pacientes fallecidos.

```python
vivos = data[data['DEATH_EVENT'] == 0]['serum_creatinine']
fallecidos = data[data['DEATH_EVENT'] == 1]['serum_creatinine']
plt.figure(figsize=(8,5))
plt.boxplot([vivos, fallecidos], labels=['Vivos', 'Fallecidos'])
plt.title("Comparación del nivel de Serum Creatinine entre vivos y fallecidos")
plt.ylabel("Serum Creatinine (mg/dL)")
plt.xlabel("Estado del paciente")

plt.show()
```

de este grafico se concluye que la falla renal medida por la creatinina cérica en una si afecta en el fallecimiento de las personas con anormalidades cardiacas estudiadas por este dataset

```python
mean_age_alive = data[data["DEATH_EVENT"] == 0]["age"].mean()
mean_age_dead = data[data["DEATH_EVENT"] == 1]["age"].mean()

# Gráfica de barras
plt.figure(figsize=(6,5))
plt.bar(["Vivos", "Fallecidos"], [mean_age_alive, mean_age_dead])
plt.title("Promedio de Edad según estado del paciente")
plt.ylabel("Edad (años)")
plt.show()
```

aca vemos otro ejemplo donde nos muestra que la edad de las personas SI influye en su muerte debido a que sus órganos o los músculos de estos pueden estar más debiles
